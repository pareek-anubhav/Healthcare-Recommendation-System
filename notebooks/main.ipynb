```python
#label encoding
X = df.drop(columns=['diseases'])
y = df['diseases']

print("Feature Matrix Shape :", X.shape)
print("Target Shape :", y.shape)


from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y)

print("Encoding Completed Successfully.")


disease_mapping = pd.DataFrame({
    "Encoded Label": range(len(label_encoder.classes_)),
    "Disease": label_encoder.classes_
})

display(disease_mapping)

import joblib

joblib.dump(label_encoder, "../models/label_encoder.pkl")

print("Label Encoder Saved Successfully.")
feature_names = X.columns.tolist()

joblib.dump(feature_names, "../models/feature_names.pkl")

print("Feature Names Saved Successfully.")
```

```python
print("="*50)
print("Feature Matrix :", X.shape)
print("Encoded Target :", y_encoded.shape)
print("Total Features :", X.shape[1])
print("Total Diseases :", len(label_encoder.classes_))
print("="*50)
```

```python
df.to_csv("../datasets/processed/clean_dataset.csv", index=False)
```

```python
from sklearn.model_selection import (
    train_test_split,
    GridSearchCV,
    cross_val_score)


X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.2,
    random_state=42,
    stratify=y_encoded
)
print(X_train.shape)
print(X_test.shape)
```

```python
# model training
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from xgboost import XGBClassifier


models = {

    "Logistic Regression":
    LogisticRegression(max_iter=1000,n_jobs=-1),

    "Decision Tree":
    DecisionTreeClassifier(random_state=42),

    "Random Forest":
    RandomForestClassifier(
        random_state=42, 
        n_jobs=-1
    ),

    "XGBoost":
    XGBClassifier(
        eval_metric="mlogloss",
        random_state=42,
        n_jobs=-1
    )

}
```

```python
#model compairing
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay)



results = []

for name, model in models.items():
    print(f"Training {name}...")
    model.fit(X_train, y_train)

    pred = model.predict(X_test)

    accuracy = accuracy_score(y_test, pred)

    results.append({
        "Model": name,
        "Accuracy": accuracy
    })

results = pd.DataFrame(results)

results.sort_values(
    by="Accuracy",
    ascending=False
)
```

```python
# ======================================================
# Step 2.8 : Select Best Model
# ======================================================

best_model = models["Logistic Regression"]

# Train on the training data
best_model.fit(X_train, y_train)

print("Selected Model : Logistic Regression")
```

```python
# ======================================================
# Step 2.9 : Model Evaluation
# ======================================================

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

y_pred = best_model.predict(X_test)

accuracy = accuracy_score(y_test,y_pred)

print(f"Accuracy : {accuracy:.4f}")

print(classification_report(
    y_test,
    y_pred,
    target_names=label_encoder.classes_
))
```

```python
# ======================================================
# Step 2.11 : Cross Validation
# ======================================================

from sklearn.model_selection import cross_val_score

scores = cross_val_score(
    best_model,
    X,
    y_encoded,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)

print("Fold Accuracy :")
print(scores)

print()

print("Average Accuracy :",
      scores.mean())
```

```python
# ======================================================
# Step 2.12 : Important Symptoms
# ======================================================

importance = np.mean(
    np.abs(best_model.coef_),
    axis=0
)

importance_df = pd.DataFrame({
    "Symptom":X.columns,
    "Importance":importance
})

importance_df = importance_df.sort_values(
    by="Importance",
    ascending=False
)

importance_df.head(20)
```

```python
plt.figure(figsize=(10,8))

sns.barplot(
    data=importance_df.head(20),
    x="Importance",
    y="Symptom"
)

plt.title("Top 20 Important Symptoms")

plt.tight_layout()

plt.show()
```

```python
# ======================================================
# Step 2.13 : Prediction
# ======================================================

sample = X_test.iloc[[0]]

prediction = best_model.predict(sample)

disease = label_encoder.inverse_transform(prediction)

print("Predicted Disease :",disease[0])
```

```python
# ======================================================
# Standardize Disease Names
# ======================================================

datasets = [description, medications, diets, precautions, workout]

for df in datasets:

    df["disease"] = (
        df["disease"]
        .astype(str)
        .str.lower()
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)
    )
```

```python
# ======================================================
# Step 3.11 : Create Master Recommendation Dataset
# ======================================================

recommendation_df = description.merge(
    medications,
    on="disease",
    how="left"
)

recommendation_df = recommendation_df.merge(
    diets,
    on="disease",
    how="left"
)

recommendation_df = recommendation_df.merge(
    precautions,
    on="disease",
    how="left"
)

recommendation_df = recommendation_df.merge(
    workout,
    on="disease",
    how="left"
)

recommendation_df.head()
```

```python
print(recommendation_df.shape)

recommendation_df.info()

recommendation_df.head()
```

```python
recommendation_df.to_csv(
    "../models/healthcare_recommendation_database.csv",
    index=False
)

print("Healthcare Recommendation Database Created Successfully!")
```

```python
# ======================================================
# Step 3.14 : Healthcare Recommendation Function
# ======================================================

def healthcare_recommendation(disease):

    disease = disease.lower().strip()

    result = recommendation_df[
        recommendation_df["disease"] == disease
    ]

    if result.empty:
        return None

    return result.iloc[0]
```

```python
predicted_disease = "panic disorder"

recommendation = healthcare_recommendation(
    predicted_disease
)

recommendation
```
